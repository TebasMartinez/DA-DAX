```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
DAX = pd.read_csv("data/DAX.csv")
EURtoGBP = pd.read_csv("data/EURtoGBP.csv")
EURtoUSD = pd.read_csv("data/EURtoUSD.csv")
close_prices = pd.read_csv("data/close_prices.csv")
interest_rates = pd.read_csv("data/interest_rates.csv")
monthly_close_prices = pd.read_csv("data/monthly_close_prices.csv")
monthly_returns = pd.read_csv("data/monthly_returns.csv")
monthly_volumes = pd.read_csv("data/monthly_volumes.csv")
returns = pd.read_csv("data/returns.csv")
unemployment = pd.read_csv("data/unemployment.csv")
volumes = pd.read_csv("data/volumes.csv")
merged_daily_market = pd.read_csv("data/merged_daily_market.csv")
merged_monthly_market = pd.read_csv("data/merged_monthly_market.csv")
```

```python
merged_daily_market.head()
```

## Daily df correlations

### Per company

```python
for company in set(merged_daily_market['company']):
    print(f"{company}:")
    sns.scatterplot(data=merged_daily_market[merged_daily_market['company']==company], x='close', y='1eur_usd')
    plt.show();
```

```python
for company in set(merged_daily_market['company']):
    print(f"{company}:")
    sns.scatterplot(data=merged_daily_market[merged_daily_market['company']==company], x='close', y='1eur_gbp')
    plt.show();
```

```python
for company in set(merged_daily_market['company']):
    df_numerical = merged_daily_market[merged_daily_market['company']==company][['close', 'return', 'volume', '1eur_usd', '1eur_gbp']]
    correlation_matrix = df_numerical.corr(method='spearman')
    mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
    plt.figure(figsize=(7, 5))
    sns.heatmap(correlation_matrix, mask=mask, annot=True, cmap="coolwarm")
    plt.title(f"Correlation Heatmap for {company}")
    plt.show()
```
